```python

# Separate features (X) and target (y)

X = df.drop(columns=["ID_code", "target"])
y = df["target"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting Target Distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature Scaling Completed!")
print("Training Shape:", X_train_scaled.shape)
print("Testing Shape:", X_test_scaled.shape)
```

```python

from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model_lr.fit(X_train_scaled, y_train)

print("Logistic Regression Model Trained Successfully!")
```

```python

y_pred_lr = model_lr.predict(X_test_scaled)

print("First 10 Predictions:")
print(y_pred_lr[:10])
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("Accuracy:", accuracy_score(y_test, y_pred_lr))
print("Precision:", precision_score(y_test, y_pred_lr))
print("Recall:", recall_score(y_test, y_pred_lr))
print("F1 Score:", f1_score(y_test, y_pred_lr))
print("ROC-AUC:", roc_auc_score(
    y_test, model_lr.predict_proba(X_test_scaled)[:, 1]
))
```

```python

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred_lr)

print("Confusion Matrix:")
print(cm)

display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Transaction", "Transaction"]
)

display.plot(cmap="Blues")
plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

```python

model_lr_balanced = LogisticRegression(
    max_iter=1000,
    random_state=42,
    class_weight="balanced"
)

model_lr_balanced.fit(X_train_scaled, y_train)

y_pred_balanced = model_lr_balanced.predict(X_test_scaled)

print("Balanced Logistic Regression Results")
print("Accuracy:", accuracy_score(y_test, y_pred_balanced))
print("Precision:", precision_score(y_test, y_pred_balanced))
print("Recall:", recall_score(y_test, y_pred_balanced))
print("F1 Score:", f1_score(y_test, y_pred_balanced))
print("ROC-AUC:", roc_auc_score(
    y_test,
    model_lr_balanced.predict_proba(X_test_scaled)[:, 1]
))
```

```python

import pandas as pd
import numpy as np

df = pd.read_csv("dataset/train(1).csv")

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)
```

```python

X = df.drop(columns=["ID_code", "target"])
y = df["target"]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

y_pred_rf = model_rf.predict(X_test)

print("Random Forest Results")
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall:", recall_score(y_test, y_pred_rf))
print("F1 Score:", f1_score(y_test, y_pred_rf))
print("ROC-AUC:", roc_auc_score(
    y_test,
    model_rf.predict_proba(X_test)[:, 1]
))
```

##  Model Comparison and Selection

Four machine learning models were developed and evaluated:

1. Logistic Regression
2. Balanced Logistic Regression
3. Random Forest
4. XGBoost

### Model Evaluation

All models were evaluated using Accuracy, Precision, Recall, F1 Score, and ROC-AUC.

### Best Model Selection

XGBoost achieved the highest F1 Score of **43.74%** among the four models.

It achieved:
- Accuracy: 82.30%
- Precision: 32.13%
- Recall: 68.48%
- F1 Score: 43.74%
- ROC-AUC: 84.90%

Therefore, XGBoost was selected as the current best candidate because it provides the best F1 Score and a reasonable balance between Precision and Recall.

Balanced Logistic Regression achieved the highest Recall of 77.54%, making it another important model for identifying potential banking customers.

### Conclusion

XGBoost is selected as the preferred model for further validation and future customer transaction prediction.

## Challenges Faced

During the Customer Transaction Prediction project, the following challenges were encountered:

1. **Class Imbalance:** The dataset contained approximately 89.95% non-transaction customers and 10.05% transaction customers.

2. **Feature Complexity:** The dataset contained 200 anonymized features, making feature interpretation difficult.

3. **Low Recall:** The initial Logistic Regression model had low recall, meaning many actual transaction customers were incorrectly classified.

4. **Model Performance:** Different machine learning algorithms produced different results, requiring careful model comparison.

5. **Model Selection:** Accuracy alone was not sufficient to select the best model because of the imbalanced dataset.

### Solutions Implemented

- Used stratified train-test splitting.
- Applied class weighting in Balanced Logistic Regression.
- Used `scale_pos_weight` in XGBoost.
- Compared models using Precision, Recall, F1 Score, and ROC-AUC.
- Selected XGBoost as the current best candidate based on F1 Score.

## Final Conclusion

The Customer Transaction Prediction project aimed to predict whether a banking customer would make a transaction in the future.

The dataset contained 200,000 customer records and 200 anonymized features. Initial analysis identified a class imbalance, with approximately 89.95% non-transaction customers and 10.05% transaction customers.

Four machine learning algorithms were implemented and evaluated:

- Logistic Regression
- Balanced Logistic Regression
- Random Forest
- XGBoost

Among these models, XGBoost achieved the highest F1 Score of 43.74%, with a Recall of 68.48% and ROC-AUC of 84.90%.

Therefore, XGBoost was selected as the preferred model based on the F1 Score. Balanced Logistic Regression also showed strong performance in identifying transaction customers, achieving 77.54% Recall.

The selected XGBoost model and StandardScaler were saved for future use.

This project provided practical experience in data preprocessing, handling imbalanced datasets, machine learning model development, model evaluation, and model comparison.
